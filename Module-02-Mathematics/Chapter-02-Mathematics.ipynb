## 1. Linear Algebra

---

### In Simple English

**Linear Algebra** is the mathematics of organizing numbers into geometric grids and transforming them together in bulk.

Instead of writing code that adds or multiplies one number at a time inside millions of nested loops, linear algebra packages entire datasets, model parameters, and layer activations into structured blocks called **tensors**.

Think of it this way: If a standard calculator handles one recipe ingredient at a time, linear algebra is a factory assembly line that transforms 10,000 cakes simultaneously in a single stroke.

The key point is: **Every modern neural network is essentially a sequence of linear transformations (matrix multiplications) separated by non-linear activation functions.**

---

### Real-Life Example

**Scenario:** Building a movie recommendation system (like Netflix).

| Real Life | Tech Equivalent | Math Representation |
|-----------|----------------|---------------------|
| Single movie rating (e.g. 4.5 stars) | Scalar | 0-D Tensor: `x = 4.5` |
| One user's ratings across 5 genres (Action, Comedy, Romance, Sci-Fi, Drama) | Vector | 1-D Tensor: `v = [5, 2, 1, 4, 3]` |
| All 10,000 users' ratings across 5 genres | Matrix | 2-D Tensor: Shape `(10000, 5)` |
| Batch of 32 movie posters (images with height, width, and 3 RGB channels) | 4-D Tensor | Shape `(32, 3, 224, 224)` |
| Measuring how closely a user's taste matches a movie's genre tags | Dot Product | `user_vector · movie_vector` |
| Finding the top 2 underlying taste dimensions across all users | Eigenvectors / PCA | Principal axes of covariance matrix |

**Why this matters:** GPUs (Graphics Processing Units) are hardware chips engineered specifically to compute billions of matrix multiplications per second. Understanding tensor shapes and operations allows you to debug dimensional mismatches, which account for over 80% of deep learning runtime bugs.

---

### How It Works (Step by Step)

#### 1.1 Scalars, Vectors, Matrices, and Tensors
- **Scalar (Rank 0):** A single standalone number (e.g. learning rate $\alpha = 0.001$, loss value $L = 0.245$).
- **Vector (Rank 1):** An ordered list of numbers with 1 dimension (e.g. a word embedding vector of length 768, a bias vector $b$).
- **Matrix (Rank 2):** A 2-dimensional grid of numbers with rows and columns (e.g. weight matrix $W$ with shape `(in_features, out_features)`, or a batch of tabular features `(batch_size, num_features)`).
- **Tensor (Rank $N$):** The generalized container for $N$ dimensions.
  - **Rank 3:** Sequential or text data: `(batch_size, sequence_length, embedding_dim)`
  - **Rank 4:** Computer vision batches: `(batch_size, channels, height, width)`
  - **Rank 5:** Video batches: `(batch_size, frames, channels, height, width)`

#### 1.2 Matrix Multiplication — The Engine of Neural Networks
A linear layer computes:
$$Y = XW + b$$
where $X$ is the input matrix, $W$ is the weight matrix, and $b$ is the bias vector.
- **Dimension Rule:** To multiply matrix $A$ of shape $(M \times K)$ by matrix $B$ of shape $(K \times N)$, the inner dimensions **must match** ($K = K$). The resulting matrix has shape $(M \times N)$.
- Each element $(i, j)$ in the output is the dot product of row $i$ of $A$ and column $j$ of $B$.

#### 1.3 Dot Products & Geometric Meaning
The dot product of two vectors $u$ and $v$ is:
$$u \cdot v = \sum_{i=1}^n u_i v_i = \|u\| \|v\| \cos(\theta)$$
- If two vectors point in the **exact same direction**, $\cos(0^\circ) = 1$, yielding the maximum positive dot product.
- If two vectors are **orthogonal (perpendicular)**, $\cos(90^\circ) = 0$, dot product is 0 (no shared information).
- If they point in **opposite directions**, $\cos(180^\circ) = -1$, yielding a negative dot product.
- **Cosine Similarity:** Normalized dot product $\frac{u \cdot v}{\|u\| \|v\|}$, used in vector databases, semantic search, and the Query-Key attention mechanism in Transformers ($QK^T$).

#### 1.4 Transpose, Inverse, and Identity Matrix
- **Transpose ($A^T$):** Flips a matrix over its diagonal, turning rows into columns. If $A$ has shape $(M \times N)$, $A^T$ has shape $(N \times M)$. Crucial in backpropagation to align dimensions when propagating gradients backwards.
- **Identity Matrix ($I$):** A square matrix with $1$s on the diagonal and $0$s elsewhere. Multiplying any vector or matrix by $I$ leaves it unchanged ($A I = A$). Acts as the number "1" for matrices.
- **Inverse ($A^{-1}$):** A matrix such that $A A^{-1} = I$. Only exists for non-singular square matrices.
  - *Why Deep Learning avoids explicit inverses:* Computing $A^{-1}$ costs $O(N^3)$ compute and is numerically unstable due to floating-point rounding. Neural networks use iterative gradient descent instead of matrix inversion to find optimal parameters.

#### 1.5 Eigenvalues and Eigenvectors (Conceptual PCA)
When a matrix $A$ multiplies most vectors, it changes both their length and their direction. However, certain special vectors only get scaled (stretched or compressed) without changing their spatial direction:
$$A v = \lambda v$$
- $v$ is the **eigenvector**.
- $\lambda$ is the **eigenvalue** (the scaling factor).
- **Why PCA needs them:** Principal Component Analysis (PCA) finds the eigenvectors of the data covariance matrix. The eigenvector with the largest eigenvalue points in the direction of greatest data variance, allowing us to compress high-dimensional inputs down to fewer dimensions while preserving maximum information.

#### 1.6 Norms (L1, L2) in Loss Functions and Regularization
Norms measure the length or magnitude of a vector:
- **L1 Norm (Manhattan Norm):**
  $$\|x\|_1 = \sum_{i=1}^n |x_i|$$
  - Sum of absolute values. Drives parameters to **exact zero**, creating sparse models (used in L1 / Lasso regularization).
- **L2 Norm (Euclidean Norm):**
  $$\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2}$$
  - Direct straight-line distance. Penalizes large weights heavily because of squaring, keeping weights small, smooth, and distributed (used in L2 / Ridge regularization, also known as **Weight Decay**).

---

### Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Tensor Rank & Shape** | Number of dimensions and size along each dimension | Mismatched tensor shapes are the #1 source of model bugs |
| **Matrix Multiplication** | Combining rows of input with columns of weights | Forms the core forward pass computation of every dense, convolutional, and attention layer |
| **Dot Product** | Sum of element-wise products of two equal-length vectors | Measures directional alignment and semantic similarity |
| **Cosine Similarity** | Normalized dot product ranging from -1 to +1 | Evaluates text embeddings and computes attention scores |
| **Matrix Transpose** | Swapping rows and columns ($A^T$) | Aligns shapes during backpropagation gradient computation |
| **L2 Norm (Weight Decay)** | Euclidean length of weight vector | Prevents overfitting by penalizing excessively large weights |
| **L1 Norm (Lasso)** | Sum of absolute magnitudes | Encourages weight sparsity for feature selection |

### Common Misconceptions

| Myth | Reality |
|------|---------|
| Matrix multiplication ($A @ B$) is just element-wise multiplication ($A * B$). | False. Element-wise multiplication requires identical shapes and multiplies matching cells (Hadamard product). Matrix multiplication computes row-column dot products and requires inner dimensions to match. |
| You must invert matrices ($A^{-1}$) to solve neural network equations. | False. Computing matrix inverses is $O(N^3)$ and numerically brittle. Neural networks use iterative first-order gradient descent instead. |
| Tensors are only useful for images and computer vision. | False. Text (token embeddings), audio spectrograms, graph node features, and time series are all structured as 2D, 3D, or 4D tensors. |

---

### When to Use This

- Check tensor shapes immediately whenever defining a new layer or reshaping data (`x.shape`, `x.view()`, `x.reshape()`).
- Use **dot product / cosine similarity** for semantic retrieval, recommendation scoring, and attention weights.
- Apply **L2 Regularization (Weight Decay)** as your default regularizer to prevent overfitting.
- Apply **L1 Regularization** when you explicitly desire sparse activations or feature selection.

| **Matrix Product** | Inner product of rows and columns | `(128, 512) @ (512, 64) -> (128, 64)` |
| **Dot Product** | Scalar sum of pairwise products | `[1, 2] · [3, 4] = 1*3 + 2*4 = 11` |

````python
# Code Demo: Linear Algebra Foundations for Deep Learning
# Concept: Tensors, Matrix Multiplication, Dot Products, Transpose & Norms
# -----------------------------------------------------------------------

# 1. Scalars, Vectors, Matrices, and Higher-Order Tensors
scalar = np.array(3.14159)                             # Rank-0 tensor (scalar)
vector = np.array([1.5, -2.0, 3.5])                    # Rank-1 tensor (vector)
matrix = np.array([[1.0, 2.0], [3.0, 4.0]])            # Rank-2 tensor (matrix)
tensor_4d = np.random.randn(8, 3, 32, 32)              # Rank-4 tensor: 8 images, 3 channels, 32x32 pixels

print("1. Tensor Shapes and Dimensionalities:")
print(f"   Scalar shape:    {scalar.shape}  | Rank: {scalar.ndim}")
print(f"   Vector shape:    {vector.shape}     | Rank: {vector.ndim}")
print(f"   Matrix shape:    {matrix.shape}     | Rank: {matrix.ndim}")
print(f"   4D Image batch:  {tensor_4d.shape} | Rank: {tensor_4d.ndim}")

# 2. Matrix Multiplication: Forward Pass of a Dense Layer (Y = X @ W + b)
batch_size = 4
in_features = 3
out_features = 2

np.random.seed(42)
X = np.random.randn(batch_size, in_features)           # Input batch: shape (4, 3)
W = np.random.randn(in_features, out_features)         # Weight matrix: shape (3, 2)
b = np.array([0.1, -0.2])                              # Bias vector: shape (2,)

# Compute linear transformation
Y = np.matmul(X, W) + b                                # Equivalent to X @ W + b

print("\n2. Matrix Multiplication Forward Pass (Y = XW + b):")
print(f"   X shape: {X.shape} @ W shape: {W.shape} -> Y shape: {Y.shape}")
print(f"   Output activations:\n{Y}")

# 3. Dot Product and Cosine Similarity (Semantic Alignment)
v1 = np.array([1.0, 2.0, 3.0])
v2 = np.array([2.0, 4.0, 6.0])                         # Exactly parallel to v1 (scaled by 2)
v3 = np.array([-3.0, 0.0, 1.0])                        # Orthogonal to v1 (dot product = 0)

dot_v1_v2 = np.dot(v1, v2)
dot_v1_v3 = np.dot(v1, v3)
cos_sim_12 = dot_v1_v2 / (np.linalg.norm(v1) * np.linalg.norm(v2))
cos_sim_13 = dot_v1_v3 / (np.linalg.norm(v1) * np.linalg.norm(v3))

print("\n3. Dot Products & Cosine Similarities:")
print(f"   dot(v1, v2) = {dot_v1_v2:.2f} | Cosine Similarity (Parallel):   {cos_sim_12:.4f}")
print(f"   dot(v1, v3) = {dot_v1_v3:.2f} | Cosine Similarity (Orthogonal): {cos_sim_13:.4f}")

# 4. Transpose and Identity Matrix
M = np.array([[1, 2, 3], [4, 5, 6]])                   # Shape (2, 3)
M_T = M.T                                              # Shape (3, 2)
I = np.eye(3)                                          # Identity matrix (3, 3)

print("\n4. Transpose and Identity:")
print(f"   Original shape: {M.shape} -> Transposed shape: {M_T.shape}")
print(f"   Matrix times Identity equals itself: {np.allclose(M @ I, M)}")

# 5. Vector Norms (L1 and L2 Regularization / Distance)
weights = np.array([0.5, -1.2, 2.0, -0.05])

l1_norm = np.sum(np.abs(weights))                      # L1 Manhattan norm (Lasso)
l2_norm = np.sqrt(np.sum(weights ** 2))                # L2 Euclidean norm (Ridge)

print("\n5. Vector Norms for Regularization:")
print(f"   Weight vector: {weights}")
print(f"   L1 Norm (Sum of absolute values): {l1_norm:.4f}")
print(f"   L2 Norm (Euclidean length):       {l2_norm:.4f}")
````

## 2. Calculus & Differentiation

---

### In Simple English

**Calculus** is the mathematics of change and sensitivity.

When a neural network makes a wrong prediction, the error is quantified by a single number called the **Loss**. Deep learning models contain millions or billions of adjustable parameters (weights). 

Calculus answers the single most important question in training:
> *"If I tweak this specific weight by a tiny hair, will the model's loss go up or down, and by how much?"*

Think of it this way: You are hiking down a mountain covered in dense fog. You cannot see the base of the valley. But by feeling the tilt of the ground right beneath your feet, you know which direction steps downward.

The key point is: **The derivative tells us the slope, the gradient combines all the slopes into a vector pointing uphill, and gradient descent steps in the exact opposite direction to minimize error.**

---

### Real-Life Example

**Scenario:** Adjusting shower temperature.

| Real Life | Tech Equivalent | Math Representation |
|-----------|----------------|---------------------|
| Water temperature is too cold | High model error (Loss) | Loss $L = (T_{\text{actual}} - T_{\text{target}})^2$ |
| Nudging the hot water knob 1 millimeter | Changing weight $w_{\text{hot}}$ by $\Delta w$ | Small perturbation $\Delta w$ |
| How much the temperature shifts per millimeter turned | Partial derivative of temperature w.r.t hot knob | $\frac{\partial T}{\partial w_{\text{hot}}}$ |
| Adjusting both hot and cold knobs at the same time | Full gradient vector of the system | $\nabla L = \left[ \frac{\partial L}{\partial w_{\text{hot}}}, \frac{\partial L}{\partial w_{\text{cold}}} \right]$ |
| Knob moves valve $\rightarrow$ valve moves water flow $\rightarrow$ water flow changes temperature | Chain of dependent operations | Chain Rule: $\frac{\partial T}{\partial \text{knob}} = \frac{\partial T}{\partial \text{flow}} \cdot \frac{\partial \text{flow}}{\partial \text{valve}} \cdot \frac{\partial \text{valve}}{\partial \text{knob}}$ |

**Why this matters:** In a neural network with 100 billion parameters, guessing weights randomly would take longer than the age of the universe. Calculus provides a direct mathematical formula to adjust all 100 billion parameters simultaneously in milliseconds.

---

### How It Works (Step by Step)

#### 2.1 What a Derivative Actually Means
The derivative of a function $f(x)$ measures its instantaneous rate of change:
$$\frac{df}{dx} = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$
- If $\frac{df}{dx} > 0$: Increasing $x$ increases $f(x)$. (Slope is positive)
- If $\frac{df}{dx} < 0$: Increasing $x$ decreases $f(x)$. (Slope is negative)
- If $\frac{df}{dx} = 0$: The function is flat (a local minimum, local maximum, or inflection point).

#### 2.2 Partial Derivatives — Sensitivity to Each Weight
Neural networks do not have just 1 input; they have millions of weights: $L(w_1, w_2, \dots, w_n)$.
- A **partial derivative** $\frac{\partial L}{\partial w_i}$ measures how the loss changes when varying $w_i$ while holding **all other weights completely frozen**.
- The symbol $\partial$ ("del") indicates that other variables are treated as constants during the differentiation.

#### 2.3 The Chain Rule — The Mathematical Heartbeat of Backpropagation
A deep neural network is a composite function of nested operations:
$$\hat{y} = f_3(f_2(f_1(x)))$$
If $z$ depends on $y$, and $y$ depends on $x$:
$$\frac{\partial z}{\partial x} = \frac{\partial z}{\partial y} \cdot \frac{\partial y}{\partial x}$$
- **Why this is profound:** To find the gradient of the loss with respect to a weight deep in layer 1, we do not need to recompute everything from scratch. We simply compute **local derivatives** at each layer and multiply them together as error signals flow backward from output to input. This backward propagation of gradients is called **Backpropagation**.

#### 2.4 Gradients — Direction of Steepest Ascent / Descent
The **gradient** $\nabla L$ (pronounced "nabla L") is the vector that collects all individual partial derivatives:
$$\nabla_W L = \begin{bmatrix} \frac{\partial L}{\partial w_1} \\ \frac{\partial L}{\partial w_2} \\ \vdots \\ \frac{\partial L}{\partial w_n} \end{bmatrix}$$
- **Fundamental Geometric Property:** The gradient vector points in the direction of **steepest increase** (steepest ascent) of the function.
- **Gradient Descent Update Rule:** To *minimize* the loss, we take a step in the exact **opposite direction** (steepest descent) scaled by a step size called the **learning rate** ($\eta$):
$$W_{\text{new}} = W_{\text{old}} - \eta \cdot \nabla_W L$$

#### 2.5 Computational Graphs & PyTorch Autograd
How do modern frameworks like PyTorch and JAX compute gradients for arbitrary user code?
- **Forward Pass:** The framework builds a **Directed Acyclic Graph (DAG)** where leaf nodes are input tensors and weights, intermediate nodes are mathematical operations ($+, \times, \text{ReLU}$), and the root node is the scalar Loss.
- **Backward Pass (`loss.backward()`):** The framework traverses the DAG in reverse order (Reverse-Mode Automatic Differentiation). It calls the pre-programmed local derivative of each node and multiplies it by incoming gradients using the chain rule.

---

### Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Derivative** | Rate of change / slope of a function at a specific point | Tells us whether nudging a variable increases or decreases output |
| **Partial Derivative** | Slope with respect to one variable while holding others fixed | Measures the individual contribution of each weight to total error |
| **Chain Rule** | Derivative of composite functions: $\frac{\partial z}{\partial x} = \frac{\partial z}{\partial y} \cdot \frac{\partial y}{\partial x}$ | Enables backpropagation across arbitrary network depths |
| **Gradient Vector ($\nabla$)** | Vector of all partial derivatives | Dictates the direction of parameter updates |
| **Learning Rate ($\eta$)** | Hyperparameter controlling step size | Too small = slow training; too large = divergence / instability |
| **Computational Graph** | Directed graph tracking forward operations | Foundation of PyTorch autograd engine |

---

### Visual Overview

```

  [ Input x ] --+
                |--> [ Node: z = w * x + b ] --> [ Node: a = ReLU(z) ] --> [ Loss: L = (a - y)^2 ]
  [ Weight w ] -+

Backward Pass / Backpropagation (Gradients Flow Left):
  [ dL/dw = dL/da * da/dz * dz/dw ] <-- [ dL/dz = dL/da * da/dz ] <-- [ dL/da = 2*(a - y) ]
            ^
            |
    Chain Rule multiplies local derivatives backwards step by step!

Gradient Descent Step:
    Loss
     |      \
     |       \
     |        *  <- Current weights (High Loss)
     |         \
     |          \  Direction of -Gradient (-dL/dw)
     |           \----->
     |                 *  <- Updated weights (Lower Loss!)
     +---------------------------------------- Weights (w)
```

---

### Common Misconceptions

| Myth | Reality |
|------|---------|
| PyTorch uses symbolic calculus (like Wolfram Alpha or SymPy) to derive giant algebraic equations. | False. PyTorch uses **Automatic Differentiation (Autograd)**, which numerically evaluates local derivatives at each node during runtime. |
| A gradient of zero always means you found the best possible global minimum. | False. In high-dimensional deep learning, zero gradients usually occur at **saddle points** or shallow plateaus rather than true global minima. |
| Making the learning rate as small as possible is always the safest option. | False. If the learning rate is too small, training gets stuck in local plateaus, takes months to converge, or overfits early. |

---

### When to Use This

- Debugging exploding or vanishing gradients (gradients become `NaN` or `0.0` due to too many chained multiplications).
- Designing custom loss functions or custom neural network layers where you must understand the gradient flow.
- Tuning learning rate schedules (e.g. Cosine Annealing, Warmup) to ensure steady descent toward low loss valleys.

---

### Section Quick Reference

| Term | Definition | Formula / Rule |
|------|-----------|----------------|
| **Derivative** | Slope of tangent line | $f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$ |
| **Chain Rule** | Derivative of composition | $\frac{dz}{dx} = \frac{dz}{dy} \cdot \frac{dy}{dx}$ |
| **Gradient** | Vector of partial derivatives | $\nabla_W L = [\frac{\partial L}{\partial w_1}, \dots, \frac{\partial L}{\partial w_n}]^T$ |
| **Update Step** | Gradient descent step | $W \leftarrow W - \eta \cdot \nabla_W L$ |

````python
# Code Demo: Calculus, Chain Rule & Autograd Engine from Scratch
# Concept: Numerical vs Analytical Derivatives, Chain Rule & Gradient Descent
# ---------------------------------------------------------------------------

# 1. Numerical Derivative vs Analytical Derivative
# Function: f(x) = x^2 + 3x
def f(x):
    return x**2 + 3*x

# Analytical derivative: f'(x) = 2x + 3
def f_prime_analytical(x):
    return 2*x + 3

# Numerical derivative using finite difference: [f(x+h) - f(x-h)] / (2h)
def f_prime_numerical(x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

test_x = 4.0
analytical_result = f_prime_analytical(test_x)
numerical_result = f_prime_numerical(test_x)

print("1. Derivative Verification for f(x) = x^2 + 3x at x = 4.0:")
print(f"   Analytical slope (2x + 3): {analytical_result:.6f}")
print(f"   Numerical slope (approx): {numerical_result:.6f}")
print(f"   Absolute discrepancy:     {abs(analytical_result - numerical_result):.2e}")

# 2. Step-by-Step Chain Rule & Backward Pass on a Mini Computational Graph
# Graph:
#   x = 2.0 (input), w = 3.0 (weight), b = -1.0 (bias)
#   Linear transformation: z = w * x + b
#   Activation: a = ReLU(z) = max(0, z)
#   True target: y_true = 10.0
#   Loss: L = (a - y_true)^2

x = 2.0
w = 3.0
b = -1.0
y_true = 10.0

# --- Forward Pass ---
z = w * x + b                         # Linear sum: 3.0 * 2.0 + (-1.0) = 5.0
a = max(0.0, z)                       # ReLU activation: 5.0
loss = (a - y_true) ** 2               # Squared error loss: (5.0 - 10.0)^2 = 25.0

print(f"\n2. Forward Pass on Computational Graph:")
print(f"   Input x: {x} | Weight w: {w} | Bias b: {b}")
print(f"   Intermediate z (w*x + b): {z:.2f}")
print(f"   Activation a (ReLU):      {a:.2f}")
print(f"   Target y_true:            {y_true:.2f}")
print(f"   Initial Loss:             {loss:.4f}")

# --- Backward Pass (Applying Chain Rule Step by Step) ---
# Step A: Local derivative of Loss w.r.t activation a
# dL/da = 2 * (a - y_true)
dL_da = 2.0 * (a - y_true)            # 2 * (5.0 - 10.0) = -10.0

# Step B: Local derivative of ReLU activation a w.r.t z
# da/dz = 1.0 if z > 0 else 0.0
da_dz = 1.0 if z > 0.0 else 0.0       # 1.0

# Step C: Local derivative of z w.r.t weight w
# dz/dw = x
dz_dw = x                             # 2.0

# Step D: Apply Chain Rule to find dL/dw
# dL/dw = (dL/da) * (da/dz) * (dz/dw)
dL_dw = dL_da * da_dz * dz_dw         # -10.0 * 1.0 * 2.0 = -20.0

# Similarly, local derivative w.r.t bias b: dz/db = 1.0
dz_db = 1.0
dL_db = dL_da * da_dz * dz_db         # -10.0 * 1.0 * 1.0 = -10.0

print(f"\n3. Backward Pass Gradients via Chain Rule:")
print(f"   dL/da (Loss sensitivity to activation): {dL_da:.2f}")
print(f"   da/dz (ReLU sensitivity to input):      {da_dz:.2f}")
print(f"   dz/dw (Linear sensitivity to weight):   {dz_dw:.2f}")
print(f"   --> Final dL/dw gradient:               {dL_dw:.2f}")
print(f"   --> Final dL/db gradient:               {dL_db:.2f}")

# 4. Gradient Descent Optimization Step
learning_rate = 0.05
w_new = w - learning_rate * dL_dw      # w_new = 3.0 - (0.05 * -20.0) = 4.0
b_new = b - learning_rate * dL_db      # b_new = -1.0 - (0.05 * -10.0) = -0.5

# Re-evaluate Loss after 1 step
z_new = w_new * x + b_new
a_new = max(0.0, z_new)
loss_new = (a_new - y_true) ** 2

print(f"\n4. Optimization Update (Learning Rate = {learning_rate}):")
print(f"   Updated weight w: {w:.2f} -> {w_new:.2f}")
print(f"   Updated bias b:   {b:.2f} -> {b_new:.2f}")
print(f"   New Loss after 1 step: {loss_new:.4f} (Decreased from {loss:.4f}!)")
````

## 3. Probability & Statistics

---

### In Simple English

**Probability and Statistics** are the mathematical languages of uncertainty, noise, and data distribution.

Real-world data is never perfectly clean. Sensor inputs have noise, images have varying lighting conditions, and text has ambiguity. Furthermore, deep neural networks do not output rigid yes/no answers; they output **probability distributions** over possible classes (e.g. "87% cat, 11% dog, 2% horse").

Think of it this way: If classical math is a rigid ruler, probability is a weather forecast. It doesn't promise rain at 2:03 PM; it estimates an 85% probability of precipitation over the afternoon.

The key point is: **Training a deep learning model is fundamentally an exercise in fitting a parameterized probability distribution to match the observed distribution of the real world.**

---

### Real-Life Example

**Scenario:** Medical diagnostic test for a rare disease.

| Real Life | Tech Equivalent | Math Representation |
|-----------|----------------|---------------------|
| Base prevalence: 1 in 1,000 people have the disease (0.1%) | Prior Probability | $P(\text{Disease}) = 0.001$ |
| Test correctly flags sick patient 99% of the time | Likelihood | $P(\text{Positive} \mid \text{Disease}) = 0.99$ |
| Test falsely flags healthy patient 2% of the time | False Positive Rate | $P(\text{Positive} \mid \text{Healthy}) = 0.02$ |
| Actual probability patient has disease after testing positive | Posterior Probability via Bayes' Theorem | $P(\text{Disease} \mid \text{Positive}) \approx 4.7\%$ |
| Spread of healthy patient blood pressure readings | Gaussian (Normal) Distribution | $\mathcal{N}(\mu=120, \sigma^2=15^2)$ |
| Finding model parameters that make test results most likely | Maximum Likelihood Estimation (MLE) | $\arg\max_\theta P(\text{Data} \mid \theta)$ |

**Why this matters:** Bayes' Theorem reveals why a "99% accurate" test can still be wrong over 95% of the time when the prior is rare. In Deep Learning, understanding priors and posteriors prevents catastrophic errors in Bayesian networks, anomaly detection, and classification under severe class imbalance.

---

### How It Works (Step by Step)

#### 3.1 Random Variables & Probability Distributions
- **Random Variable ($X$):** A variable whose values depend on outcomes of a random phenomenon.
- **Gaussian (Normal) Distribution:** $\mathcal{N}(\mu, \sigma^2)$
  $$p(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$
  - Symmetrical bell curve centered at mean $\mu$ with spread $\sigma$.
  - Used in deep learning for **weight initialization** (e.g. He / Kaiming normal, Xavier / Glorot normal), sensor noise modeling, and the latent space of Variational Autoencoders (VAEs).
- **Bernoulli Distribution:** Models binary outcomes (0 or 1) with success probability $p$.
  - Foundation of **binary classification** (models ending in a Sigmoid activation).
- **Uniform Distribution:** All intervals of equal length have identical probability.
  - Used for random sampling, dropout mask generation, and data augmentation.

#### 3.2 Mean, Variance, and Standard Deviation
- **Mean ($\mu$):** Expected central value: $\mu = \frac{1}{N} \sum_{i=1}^N x_i$
- **Variance ($\sigma^2$):** Average squared spread around the mean: $\sigma^2 = \frac{1}{N} \sum_{i=1}^N (x_i - \mu)^2$
- **Standard Deviation ($\sigma$):** $\sqrt{\sigma^2}$ (in the original units of the data).
- **Standardization ($Z$-Score Normalization):**
  $$z = \frac{x - \mu}{\sigma}$$
  - Transforms any distribution to have $\text{mean} = 0$ and $\text{std} = 1$.
  - **Why Batch Normalization & Layer Normalization are critical:** As activations pass through dozens of layers, their mean and variance can drift (Internal Covariate Shift). Normalizing activations across batches or channels keeps gradients stable and prevents vanishing/exploding signals.

#### 3.3 Bayes' Theorem
Relates prior belief to updated posterior belief given new evidence:
$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$
- $P(A)$: **Prior** (initial belief before observing evidence).
- $P(B \mid A)$: **Likelihood** (probability of evidence given hypothesis).
- $P(B)$: **Marginal Evidence** (total probability of evidence across all possibilities).
- $P(A \mid B)$: **Posterior** (updated belief after seeing evidence).

#### 3.4 Maximum Likelihood Estimation (MLE)
Suppose we observe a dataset $D = \{x_1, x_2, \dots, x_N\}$ drawn from an unknown distribution. We want to find model parameters $\theta$ that maximize the probability of observing this exact data:
$$L(\theta) = \prod_{i=1}^N P(x_i \mid \theta)$$
- Multiplying thousands of small probabilities produces floating-point underflow ($0.00000\dots$).
- Solution: Take the natural logarithm! Because $\log$ is strictly monotonically increasing, maximizing the likelihood is identical to maximizing the **Log-Likelihood**:
$$\log L(\theta) = \sum_{i=1}^N \log P(x_i \mid \theta)$$
- In optimization, algorithms minimize functions rather than maximizing them. Thus, we negate it to get the **Negative Log-Likelihood (NLL)**:
$$\text{Loss}_{\text{NLL}}(\theta) = -\sum_{i=1}^N \log P(x_i \mid \theta)$$
- **Crucial Takeaway:** Minimizing NLL is mathematically identical to Maximum Likelihood Estimation!

#### 3.5 KL Divergence (Relative Entropy)
The **Kullback-Leibler (KL) Divergence** measures how one probability distribution $Q$ differs from a reference true distribution $P$:
$$D_{\text{KL}}(P \parallel Q) = \sum_{x} P(x) \log\left( \frac{P(x)}{Q(x)} \right)$$
- It is always non-negative: $D_{\text{KL}}(P \parallel Q) \ge 0$.
- It equals 0 if and only if $P = Q$ almost everywhere.
- **Asymmetric:** $D_{\text{KL}}(P \parallel Q) \neq D_{\text{KL}}(Q \parallel P)$. (It is a divergence, not a distance metric).
- **Where it is used in Deep Learning:**
  - In **VAEs (Variational Autoencoders)**: Forces the latent distribution of encoded vectors to match a standard normal prior $\mathcal{N}(0, I)$.
  - In **Knowledge Distillation**: Measures difference between teacher model predictions $P$ and student model predictions $Q$.

---

### Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Gaussian Distribution** | Bell curve defined by mean $\mu$ and variance $\sigma^2$ | Standard prior for weights, latent representations, and noise |
| **Bernoulli Distribution** | Probability distribution of a binary event | Underpins binary cross-entropy loss with Sigmoid outputs |
| **Z-Score Standardization** | Rescaling data to 0 mean and unit variance | Accelerates gradient descent convergence; foundation of BatchNorm |
| **Bayes' Theorem** | Updating hypothesis probability with new data | Foundational to probabilistic ML and Bayesian neural networks |
| **Log-Likelihood** | Sum of log probabilities of observations | Converts products to sums and prevents numerical underflow |
| **KL Divergence** | Measure of relative entropy between two distributions | Enforces structured latent spaces in generative models (VAEs) |

| Myth | Reality |
|------|---------|

````python
# Code Demo: Probability Distributions, Normalization, MLE & KL Divergence
# Concept: Gaussian/Bernoulli Sampling, Standardization, MLE and Relative Entropy
# -----------------------------------------------------------------------------

np.random.seed(42)

# 1. Probability Distributions: Gaussian, Bernoulli & Uniform
sample_size = 5000

# Gaussian (Normal): Mean = 10.0, Std = 2.5
gaussian_data = np.random.normal(loc=10.0, scale=2.5, size=sample_size)

# Bernoulli: Success probability p = 0.7
bernoulli_data = np.random.binomial(n=1, p=0.7, size=sample_size)

# Uniform: Values between -2.0 and +2.0
uniform_data = np.random.uniform(low=-2.0, high=2.0, size=sample_size)

print("1. Probability Distribution Samples:")
print(f"   Gaussian:  Mean = {np.mean(gaussian_data):.2f} | Std = {np.std(gaussian_data):.2f}")
print(f"   Bernoulli: Mean (Fraction of 1s) = {np.mean(bernoulli_data):.2f}")
print(f"   Uniform:   Min = {np.min(uniform_data):.2f} | Max = {np.max(uniform_data):.2f} | Mean = {np.mean(uniform_data):.2f}")

# 2. Mean, Variance, and Z-Score Standardization (BatchNorm Principle)
mean_orig = np.mean(gaussian_data)
std_orig = np.std(gaussian_data)

# Standardize: z = (x - mean) / std
standardized_data = (gaussian_data - mean_orig) / std_orig

print("\n2. Feature Standardization (Z-Score Normalization):")
print(f"   Original Data:     Mean = {mean_orig:.4f}, Std = {std_orig:.4f}")
print(f"   Standardized Data: Mean = {np.mean(standardized_data):.4f}, Std = {np.std(standardized_data):.4f}")
print(f"   --> Data successfully centered at 0 with unit variance 1.0!")

# 3. Maximum Likelihood Estimation (MLE) Demonstration
# Scenario: 10 coin flips with 7 Heads and 3 Tails. Find the most likely p.
flips = np.array([1, 1, 0, 1, 1, 0, 1, 1, 0, 1])  # 7 Heads, 3 Tails
candidate_p_values = np.linspace(0.1, 0.9, 9)

best_p = None
max_log_likelihood = -float('inf')

print("\n3. Maximum Likelihood Estimation (MLE) Grid Search:")
for p in candidate_p_values:
    # Log-likelihood for Bernoulli: sum of log(p) for 1s and log(1-p) for 0s
    log_likelihood = np.sum(flips * np.log(p) + (1 - flips) * np.log(1 - p))
    if log_likelihood > max_log_likelihood:
        max_log_likelihood = log_likelihood
        best_p = p
    print(f"   p = {p:.1f} -> Log-Likelihood: {log_likelihood:.4f}")

print(f"   --> MLE Optimal Parameter: p* = {best_p:.1f} (Matches analytical 7/10 = 0.70!)")

# 4. KL Divergence (Relative Entropy) Between Two Discrete Distributions
# Distribution P: True user preference [Drama, Comedy, Action]
P = np.array([0.50, 0.30, 0.20])

# Distribution Q: Approximate recommendation model output
Q = np.array([0.40, 0.40, 0.20])

# Distribution R: Uniformly uninformative model output
R = np.array([0.333, 0.333, 0.334])

def kl_divergence(p, q):
    # KL(P || Q) = sum(P(x) * log(P(x) / Q(x)))
    return np.sum(p * np.log(p / q))

kl_P_Q = kl_divergence(P, Q)
kl_Q_P = kl_divergence(Q, P)
kl_P_R = kl_divergence(P, R)

print("\n4. KL Divergence (Relative Entropy):")
print(f"   KL(P || Q) [Model Q approximating P]: {kl_P_Q:.6f}")
print(f"   KL(Q || P) [Demonstrating asymmetry]: {kl_Q_P:.6f}")
print(f"   KL(P || R) [Model R approximating P]: {kl_P_R:.6f} (Higher divergence = worse fit)")
print(f"   Asymmetry confirmed: KL(P || Q) != KL(Q || P): {not np.isclose(kl_P_Q, kl_Q_P)}")
````

## 4. Information Theory Basics

---

### In Simple English

**Information Theory** is the science of measuring uncertainty, surprise, and information content.

Founded by Claude Shannon in 1948, information theory asks:
> *"How much surprise is contained in a message, and how many bits do we need to store it without losing anything?"*

- If a weather forecaster says: *"It will snow in July in the Sahara Desert,"* that message carries **enormous information** because it is shocking and unlikely.
- If they say: *"The sun will rise tomorrow morning,"* that message carries **virtually zero information** because you were already 100% certain it would happen.

The key point is: **In deep learning classification, the standard loss function (Cross-Entropy Loss) is taken directly from information theory — it measures how surprised our neural network is when comparing its prediction against the actual ground-truth label.**

---

### Real-Life Example

**Scenario:** Guessing the outcome of a sports match.

| Real Life | Tech Equivalent | Math Representation |
|-----------|----------------|---------------------|
| A completely unpredictable match between two equal teams (50/50) | High uncertainty / High Entropy | $H = - (0.5 \log_2 0.5 + 0.5 \log_2 0.5) = 1.0\text{ bit}$ |
| A guaranteed match where Team A has a 99.9% chance of winning | Almost zero uncertainty / Low Entropy | $H \approx 0.01\text{ bits}$ |
| The true outcome distribution of the match (e.g. Team A wins) | True Distribution $P$ | One-hot vector: $[1.0, 0.0]$ |
| A rookie bookie's estimated odds | Model Predicted Distribution $Q$ | Softmax output: $[0.75, 0.25]$ |
| Extra penalty paid for guessing wrong odds | Cross-Entropy Loss | $H(P, Q) = -\sum P(x) \log Q(x)$ |

**Why this matters:** When a model is 99% confident in the wrong answer, Cross-Entropy produces an astronomically high loss value with a steep gradient that instantly corrects the model. Mean Squared Error (MSE), on the other hand, produces a tiny gradient that gets stuck.

---

### How It Works (Step by Step)

#### 4.1 Shannon Entropy — Measuring Uncertainty
The **Entropy** $H(P)$ of a discrete probability distribution $P$ measures its average uncertainty:
$$H(P) = -\sum_{x} P(x) \log_2 P(x)$$
- Measured in **bits** (when using base-2 log) or **nats** (when using natural $\ln$).
- **Fair coin flip ($p = 0.5$):** $H = -(0.5 \log_2 0.5 + 0.5 \log_2 0.5) = 1.0$ bit. (Maximum uncertainty for 2 outcomes).
- **Biased coin ($p = 0.99$):** $H \approx 0.08$ bits. (Low uncertainty, almost predictable).
- **Deterministic outcome ($p = 1.0$):** $H = 0$ bits. (Zero uncertainty, no surprise).

#### 4.2 Cross-Entropy — Measuring Classification Loss
Suppose the true distribution of an event is $P$, but our neural network predicts distribution $Q$.
**Cross-Entropy** measures the average number of bits needed to encode events from $P$ using the model's estimated distribution $Q$:
$$H(P, Q) = -\sum_{x} P(x) \log Q(x)$$
- There is a beautiful fundamental relationship linking **Entropy**, **Cross-Entropy**, and **KL Divergence**:
$$H(P, Q) = H(P) + D_{\text{KL}}(P \parallel Q)$$
- In supervised learning, $P$ is the true dataset label distribution, which is **constant and fixed**. Therefore, $H(P)$ is a constant!
- **Crucial Mathematical Insight:**
$$\arg\min_Q H(P, Q) = \arg\min_Q D_{\text{KL}}(P \parallel Q)$$
Minimizing Cross-Entropy loss is **mathematically identical** to minimizing the KL divergence between your model's predictions and the true data distribution!

#### 4.3 Why Cross-Entropy Equals Negative Log Likelihood
In multi-class classification, the ground-truth label for sample $i$ is represented as a **one-hot vector** $y$:
$$y = [0, 0, \dots, 1, \dots, 0]$$
where only the correct class $c$ has $y_c = 1$, and all other classes $j \neq c$ have $y_j = 0$.

Now evaluate Cross-Entropy between true label $y$ and predicted probabilities $\hat{y}$:
$$L_{\text{CE}} = -\sum_{k=1}^C y_k \log(\hat{y}_k)$$
Because $y_k = 0$ for every incorrect class, all those terms vanish completely!
$$L_{\text{CE}} = - \left( 0 + 0 + \dots + 1 \cdot \log(\hat{y}_c) + \dots + 0 \right) = -\log(\hat{y}_c)$$

Notice what this is:
- $\hat{y}_c = P(\text{Class} = c \mid X; \theta)$ is the model's assigned likelihood to the true class.
- $-\log(\hat{y}_c)$ is the **Negative Log-Likelihood (NLL)** of that observation!

**The Unified Truth:**
$$\text{Maximum Likelihood Estimation (MLE)} \equiv \text{Minimizing Negative Log Likelihood (NLL)} \equiv \text{Minimizing Cross-Entropy}$$
They are not three different algorithms; they are the exact same mathematical principle viewed from three different perspectives!

---

### Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Information Content** | $-\log_2 P(x)$: Surprise of an event | Rare events convey more information than predictable events |
| **Shannon Entropy $H(P)$** | Average uncertainty of a distribution | Highest when all outcomes are equally likely (uniform) |
| **Cross-Entropy $H(P, Q)$** | Bits needed to encode $P$ using model distribution $Q$ | Standard loss function for multi-class classification |
| **Negative Log Likelihood** | $-\log P(\text{true class} \mid X)$ | Exact equivalent of categorical cross-entropy with one-hot labels |
| **Softmax + Cross-Entropy** | Turns raw model logits into probabilities, then computes NLL | Numerically stable when combined into PyTorch's `nn.CrossEntropyLoss` |

---

### Visual Overview

```
The Grand Unified Equation of Classification:

      H(P, Q)        =       H(P)       +     D_KL(P || Q)
  [ Cross-Entropy ]     [ True Entropy ]    [ KL Divergence ]
    (What we minimize)     (Constant!)        (Distance to truth)

Because H(P) is fixed by the dataset, minimizing Cross-Entropy
forces the model's distribution Q directly toward the true distribution P!

One-Hot Target Collapse:
  True Label y:           [  0.0,   1.0,   0.0  ]   (Class 1 is correct)
  Model Prediction y_hat: [  0.1,   0.8,   0.1  ]

  Cross-Entropy = - [ 0.0 * log(0.1) + 1.0 * log(0.8) + 0.0 * log(0.1) ]
                = - log(0.8)
                = Exactly the Negative Log Likelihood (-NLL)!
```

---

### Common Misconceptions

| Myth | Reality |
|------|---------|
| Mean Squared Error (MSE) is just fine for classification problems. | False. When paired with Sigmoid/Softmax, MSE suffers from severe gradient saturation (vanishing gradients) when the model makes a confident wrong prediction. Cross-Entropy maintains strong gradient signals. |
| Cross-Entropy can be negative. | False. Since predicted probabilities $\hat{y}_k \in [0, 1]$, their logarithms are always $\le 0$. The leading negative sign ensures Cross-Entropy is always $\ge 0$. |
| You should apply `Softmax` in your model code before calling PyTorch's `nn.CrossEntropyLoss`. | False! PyTorch's `nn.CrossEntropyLoss` expects raw unnormalized **logits** and applies Log-Softmax internally using the Log-Sum-Exp trick for numerical stability. |

---

### When to Use This

- Use **Binary Cross-Entropy (BCE)** for binary classification or multi-label classification tasks (e.g. image tagging with multiple labels).
- Use **Categorical Cross-Entropy** for single-label multi-class classification tasks (e.g. ImageNet, MNIST, next-token prediction in LLMs).
- Pass raw logits directly to loss functions (`nn.CrossEntropyLoss`) rather than manual `softmax` followed by `log` to avoid numerical underflow.

---

### Section Quick Reference

| Term | Definition | Formula |
|------|-----------|---------|
| **Entropy** | Measure of uncertainty | $H(P) = -\sum P(x) \log P(x)$ |
| **Cross-Entropy** | Measure of approximation error | $H(P, Q) = -\sum P(x) \log Q(x)$ |
| **NLL Equivalence** | Cross-entropy with one-hot target | $L = -\log(\hat{y}_{\text{true}})$ |

````python
# Code Demo: Shannon Entropy, Cross-Entropy Loss & NLL Equivalence
# Concept: Measuring Uncertainty, Multi-Class Cross Entropy & Negative Log Likelihood
# ----------------------------------------------------------------------------------

# 1. Shannon Entropy: Fair Coin vs Biased Coin vs Deterministic Event
def shannon_entropy(probs):
    # Filter out zero probabilities to avoid log2(0)
    p_nonzero = probs[probs > 0]
    val = -np.sum(p_nonzero * np.log2(p_nonzero))
    return 0.0 if abs(val) < 1e-12 else val

fair_coin = np.array([0.5, 0.5])
biased_coin = np.array([0.9, 0.1])
deterministic = np.array([1.0, 0.0])
uniform_die = np.array([1/6]*6)

print("1. Shannon Entropy (in bits):")
print(f"   Fair coin [0.5, 0.5]:       {shannon_entropy(fair_coin):.4f} bits (Maximum uncertainty)")
print(f"   Biased coin [0.9, 0.1]:     {shannon_entropy(biased_coin):.4f} bits (Lower uncertainty)")
print(f"   Deterministic [1.0, 0.0]:   {shannon_entropy(deterministic):.4f} bits (Zero uncertainty)")
print(f"   Fair 6-sided die:           {shannon_entropy(uniform_die):.4f} bits (Higher uncertainty)")

# 2. Categorical Cross-Entropy Loss from Scratch
def softmax(logits):
    # Subtract max for numerical stability (Log-Sum-Exp trick)
    exp_logits = np.exp(logits - np.max(logits))
    return exp_logits / np.sum(exp_logits)

def cross_entropy_loss(y_true, y_pred_probs, eps=1e-15):
    # Clip probabilities to prevent log(0)
    y_pred_probs = np.clip(y_pred_probs, eps, 1.0 - eps)
    return -np.sum(y_true * np.log(y_pred_probs))

# Scenario: 3-class classification (Cat, Dog, Bird)
# True label: Class 1 (Dog) -> One-hot: [0, 1, 0]
y_true = np.array([0.0, 1.0, 0.0])

# Model A (Good prediction): High confidence in Dog
logits_A = np.array([1.0, 4.5, 0.5])
probs_A = softmax(logits_A)
loss_A = cross_entropy_loss(y_true, probs_A)

# Model B (Poor prediction): High confidence in Bird (wrong!)
logits_B = np.array([0.5, 1.0, 4.5])
probs_B = softmax(logits_B)
loss_B = cross_entropy_loss(y_true, probs_B)

print("\n2. Cross-Entropy Loss on Predictions:")
print(f"   True class: Dog (Index 1) -> {y_true}")
print(f"   Model A Probabilities: {probs_A.round(3)} -> Loss: {loss_A:.4f}")
print(f"   Model B Probabilities: {probs_B.round(3)} -> Loss: {loss_B:.4f} (Heavy penalty for wrong guess!)")

# 3. Demonstration: Cross-Entropy Exactly Equals Negative Log-Likelihood (NLL)
# Since y_true is one-hot with index 1 = 1.0:
# Loss = -log(prob of correct class)
target_idx = np.argmax(y_true)
nll_loss_A = -np.log(probs_A[target_idx])
nll_loss_B = -np.log(probs_B[target_idx])

print("\n3. Verification: Cross-Entropy == Negative Log Likelihood (NLL):")
print(f"   Model A: Cross-Entropy = {loss_A:.6f} | NLL = {nll_loss_A:.6f} | Match: {np.isclose(loss_A, nll_loss_A)}")
print(f"   Model B: Cross-Entropy = {loss_B:.6f} | NLL = {nll_loss_B:.6f} | Match: {np.isclose(loss_B, nll_loss_B)}")

# 4. Identity Check: Cross-Entropy = Entropy(P) + KL(P || Q)
# For natural logarithm (nats):
def kl_nats(p, q, eps=1e-15):
    q = np.clip(q, eps, 1.0)
    p_nonzero = p > 0
    return np.sum(p[p_nonzero] * np.log(p[p_nonzero] / q[p_nonzero]))

def entropy_nats(p):
    p_nonzero = p[p > 0]
    val = -np.sum(p_nonzero * np.log(p_nonzero))
    return 0.0 if abs(val) < 1e-12 else val

H_P = entropy_nats(y_true)                     # Entropy of one-hot is 0
KL_div = kl_nats(y_true, probs_A)
CE_nats = cross_entropy_loss(y_true, probs_A)

print("\n4. Fundamental Identity: H(P, Q) = H(P) + D_KL(P || Q):")
print(f"   H(P) [True Label Entropy]:   {H_P:.6f} nats")
print(f"   D_KL(P || Q) [KL Divergence]: {KL_div:.6f} nats")
print(f"   H(P) + D_KL(P || Q):         {(H_P + KL_div):.6f} nats")
print(f"   Direct Cross-Entropy:        {CE_nats:.6f} nats")
print(f"   Identity holds exactly:      {np.isclose(CE_nats, H_P + KL_div)}")
````
