## 1. Learning Paradigms: Supervised, Unsupervised & Self-Supervised

---

### 🗣️ In Simple English

Machine learning algorithms learn from data, but they differ fundamentally in **what kind of supervision** they receive:

- **Supervised Learning:** The algorithm learns with a strict teacher holding an answer key. Every training example has an input feature $X$ and a true target label $y$ (e.g. predicting house price from square footage).
- **Unsupervised Learning:** The algorithm receives raw inputs $X$ with **zero answers**. It must uncover hidden clusters, natural groupings, or lower-dimensional structures on its own (e.g. grouping customers by shopping behavior).
- **Self-Supervised Learning:** The modern foundation of Deep Learning (LLMs & Vision). The algorithm is given raw data, automatically hides a portion of it (like masking words in a sentence), and trains itself to predict the hidden part. No human labeling required!

The key point is: **Supervised learning is limited by the cost of human labeling. Self-supervised learning broke through this bottleneck by letting models train on trillions of raw web tokens without human supervision.**

---

### 🌍 Real-Life Example

**Scenario:** Learning a foreign language.

| Learning Paradigm | Everyday Analogy | Technical Input | Target ($y$) | Deep Learning Example |
|-------------------|------------------|-----------------|--------------|-----------------------|
| **Supervised** | Flashcards with English on front, Spanish on back | Features $X$ (Audio recording) | Human Label $y$ ("Hello") | Whisper Speech-to-Text |
| **Unsupervised** | Listening to conversations and noticing words sound similar | Features $X$ (Customer habits) | None (Discovers clusters) | Customer Segmentation (K-Means) |
| **Self-Supervised** | Filling in blanks in a newspaper puzzle: *"The ___ jumped over the fence."* | Input $X_{\text{masked}}$ ("The ___") | Hidden token ("cat") | GPT-4 / BERT Pre-training |

---

### 📖 How It Works (Step by Step)

1. **Supervised Learning ($y = f(X)$):**
   - **Classification:** Predicting discrete categories (Spam vs Not Spam, Cat vs Dog).
   - **Regression:** Predicting continuous quantities (House price, Stock return, Temperature).
   - Cost function penalizes difference between prediction $\hat{y}$ and true label $y$.
2. **Unsupervised Learning ($P(X)$):**
   - **Clustering:** Partitioning data into $K$ cohesive groups (K-Means, DBSCAN).
   - **Dimensionality Reduction:** Compressing high-dimensional vectors to fewer dimensions while retaining variance (PCA, t-SNE, UMAP).
3. **Self-Supervised Learning:**
   - Converts unlabeled data into a supervised task using an **automated pretext task**:
     - *Autoregressive Next-Token Prediction:* Predict word $t+1$ given words $1 \dots t$ (GPT family).
     - *Masked Language Modeling:* Corrupt 15% of tokens with `[MASK]` and predict them (BERT).
     - *Contrastive Learning:* Maximize similarity between two augmented views of the same photo while pushing different photos apart (CLIP, SimCLR).

---

### 🔑 Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Supervised Learning** | Training on paired $(X, y)$ data | High accuracy but bottlenecked by expensive human annotations |
| **Unsupervised Learning** | Finding latent structure in unannotated data | Discovers natural segments and reduces dimensionality |
| **Self-Supervised Learning** | Formulating supervised prediction tasks automatically from raw data | The core training engine powering all modern Foundation Models & LLMs |
| **Pretext Task** | Artificial task used to learn general representations | Prepares weights for rapid downstream fine-tuning |

---

### 🗺️ Visual Overview

```
The Three Core Learning Paradigms:

1. Supervised Learning:
   [ Input Features X ] + [ Human Ground-Truth y ] ---> [ Model ] ---> [ Loss = (y - y_hat) ]

2. Unsupervised Learning:
   [ Input Features X ] (No Labels!)                 ---> [ Model ] ---> [ Discovered Clusters / Components ]

3. Self-Supervised Learning (The Modern LLM Engine):
   Raw Text: "The weather today is sunny"
      |-- Split automatically:
      |   Input:  "The weather today is"
      |   Target: "sunny"                            ---> [ Model ] ---> [ Predicts Next Token ]
```

---

### 🚫 Common Misconceptions

| ❌ Myth | ✅ Reality |
|--------|-----------|
| Self-supervised learning is completely unsupervised. | It is mathematically a supervised task! The difference is that the labels are generated automatically by slicing the input data itself, rather than by human annotators. |

---

### 📊 Section Quick Reference

| Paradigm | Training Data | Primary Objective | Example Algorithm |
|----------|:-------------:|-------------------|-------------------|
| **Supervised** | $(X, y)$ | Minimize prediction error | Random Forest, ResNet, XGBoost |
| **Unsupervised** | $X$ | Discover latent clusters/manifold | K-Means, PCA, GMM |
| **Self-Supervised** | $X \rightarrow (X_{\text{view}}, X_{\text{target}})$ | Predict masked/future context | GPT, BERT, SimCLR |

````python
# ✅ Code Demo: Supervised Classification vs Unsupervised Clustering
# Concept: Comparing Decision Boundaries with Labels vs Clustering without Labels
# -------------------------------------------------------------------------------

# Generate synthetic 2D dataset: 300 samples with 3 natural clusters
np.random.seed(42)
cluster_1 = np.random.normal(loc=[-3, -3], scale=0.8, size=(100, 2))
cluster_2 = np.random.normal(loc=[3, -3], scale=0.8, size=(100, 2))
cluster_3 = np.random.normal(loc=[0, 3], scale=0.8, size=(100, 2))

X_data = np.vstack([cluster_1, cluster_2, cluster_3])
y_true_labels = np.array([0]*100 + [1]*100 + [2]*100)  # Known true labels

# 1. Supervised Learning: Logistic Regression trained with true labels y
sup_clf = LogisticRegression(max_iter=1000)
sup_clf.fit(X_data, y_true_labels)
sup_accuracy = sup_clf.score(X_data, y_true_labels)

# 2. Unsupervised Learning: K-Means discovering clusters without ANY labels
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
kmeans.fit(X_data)  # Notice: Only X is passed! No y!
cluster_assignments = kmeans.labels_

print("1. Supervised Learning (Classifier trained with known labels):")
print(f"  Training Accuracy on 3 classes: {sup_accuracy*100:.2f}%")
print(f"  Model learned explicit boundary equations separating classes 0, 1, and 2.")

print("\n2. Unsupervised Learning (K-Means trained with ZERO labels):")
print(f"  Discovered Cluster Centers (Centroids):\n{kmeans.cluster_centers_.round(2)}")
print(f"  Sample Distribution across 3 clusters: {np.bincount(cluster_assignments)}")
print("  Insight: Unsupervised clustering successfully discovered the 3 physical groupings")
print("  without ever seeing a single target answer during training!")
````

## 2. Overfitting, Underfitting & the Bias-Variance Tradeoff

---

### 🗣️ In Simple English

Every machine learning model is trying to balance between two fatal errors:
- **Underfitting (High Bias):** The model is too simple, stubborn, and inflexible. It assumes the world is a flat line when reality is a curved hill. It performs terribly on both training and test data.
- **Overfitting (High Variance):** The model is too sensitive, wild, and hyper-flexible. It connects every single noisy outlier dot in the training data. It gets a 100% score on training data, but crashes instantly when shown fresh test data.

Think of custom tailoring a jacket:
- **High Bias (Underfit):** A generic one-size-fits-all burlap sack. It fits everyone poorly.
- **High Variance (Overfit):** A jacket stitched to your exact skin shape down to the millimeter while sitting down. The second you stand up or take a breath, the seams burst open.
- **The Sweet Spot:** A tailored jacket with just enough breathing room for natural movement.

The key point is: **The Bias-Variance Tradeoff is the fundamental law of machine learning: you can never eliminate both simultaneously. Your goal is to find the sweet spot of model complexity that minimizes total test error.**

---

### 🌍 Real-Life Example

**Scenario:** Predicting real estate market prices.

| Model State | Hypothesis Complexity | Train Error | Test Error | What Happened |
|-------------|-----------------------|:-----------:|:----------:|---------------|
| **High Bias (Underfit)** | Straight line: $\text{Price} = w \cdot \text{Sqft}$ | High ($\$80\text{k}$) | High ($\$82\text{k}$) | Model failed to capture non-linear location and bedroom effects |
| **High Variance (Overfit)** | 25th-degree polynomial connecting every data point | Near Zero ($\$2\text{k}$) | Catastrophic ($\$350\text{k}$) | Model memorized eccentric seller outliers |
| **Optimal Balance** | Regularized polynomial / Gradient Boosted Tree | Low ($\$18\text{k}$) | Low ($\$20\text{k}$) | Model learned the true signal while ignoring random noise |

---

### 📖 How It Works (Step by Step)

1. **The Expected Prediction Error Decomposition:**
   For any regression model, the expected test error decomposes mathematically into:
$$\mathbb{E}[(y - \hat{f}(x))^2] = \underbrace{\text{Bias}[\hat{f}(x)]^2}_{\text{Error from erroneous assumptions}} + \underbrace{\text{Var}[\hat{f}(x)]}_{\text{Error from sensitivity to training noise}} + \underbrace{\sigma^2}_{\text{Irreducible noise in data}}$$
2. **Bias:** Difference between the average prediction of our model and the true underlying reality.
3. **Variance:** How much the model's predictions would fluctuate if trained on a completely different random training set from the same domain.
4. **The Curve Dynamics:**
   - As model complexity increases: **Bias drops**, but **Variance rises**.
   - Total test error forms a U-shaped curve. The bottom of the U is the optimal model capacity.

---

### 🔑 Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Bias** | Inability of a model to capture true underlying trends | Causes high error on both training and test data |
| **Variance** | Model sensitivity to small fluctuations in training set | Causes large gap between training and validation error |
| **Irreducible Error ($\sigma^2$)** | Pure random environmental noise in the data | The theoretical floor of performance; cannot be removed by any model |
| **Model Capacity** | Expressive flexibility of an architecture | Must match data complexity to avoid underfitting or overfitting |

---

### 🗺️ Visual Overview

```
The Bias-Variance Tradeoff:

   Error
     ^
     |    \                                   /  High Variance
     |     \    Total Test Error (U-Shape)   /   (Overfitting!)
     |      \           ______              /
     |       \_________/      \____________/
     |                 ^
     |        Optimal Model Complexity
     |
     |    \                                    (Train Error drops to 0)
     |     \_________ Bias^2 (Drops)           ....................
     +------------------------------------------------------------> Model Complexity
      Low Capacity (High Bias)                   High Capacity (High Variance)
      (Linear Model)                             (25-Layer Unregularized Net)
```

---

### 🚫 Common Misconceptions

| ❌ Myth | ✅ Reality |
|--------|-----------|
| More training data always fixes high bias. | False! If your model is a straight line, feeding it 10,000,000 curved points will still yield a straight line. High bias requires increasing model capacity or engineering non-linear features. |

---

### 📊 Section Quick Reference

| Symptom | Diagnosis | Solution |
|---------|-----------|----------|
| **High Train Error & High Test Error** | High Bias (Underfitting) | Increase model complexity, add features, reduce regularization |
| **Low Train Error & High Test Error** | High Variance (Overfitting) | Add more data, add regularization (L2/Dropout), prune features |
| **Low Train Error & Low Test Error** | Optimal Fit | Model is ready for deployment testing |

````python
# ✅ Code Demo: Polynomial Degree vs Bias-Variance Tradeoff
# Concept: Visualizing Underfitting (Degree 1), Balanced (Degree 3), and Overfitting (Degree 15)
# ------------------------------------------------------------------------------------------

# Generate noisy non-linear curve: true function y = sin(x)
np.random.seed(42)
n_points = 30
x_clean = np.linspace(0, 2 * np.pi, n_points)
y_clean = np.sin(x_clean)
y_noisy = y_clean + np.random.normal(0, 0.25, n_points)

# Split into Train (70%) and Test (30%)
x_train, x_test, y_train, y_test = train_test_split(
    x_clean.reshape(-1, 1), y_noisy, test_size=0.3, random_state=42
)

degrees = [1, 3, 15]
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
titles = ["Degree 1 (Underfitting / High Bias)", "Degree 3 (Optimal Fit)", "Degree 15 (Overfitting / High Variance)"]

x_plot = np.linspace(0, 2 * np.pi, 200).reshape(-1, 1)

print("Polynomial Degree Performance Benchmark:")
print(f"  {'Degree':<8} | {'Train MSE':<12} | {'Test MSE':<12} | Diagnosis")
print("-" * 55)

for i, deg in enumerate(degrees):
    poly = PolynomialFeatures(degree=deg)
    X_train_poly = poly.fit_transform(x_train)
    X_test_poly = poly.transform(x_test)
    X_plot_poly = poly.transform(x_plot)
    
    model = LinearRegression()
    model.fit(X_train_poly, y_train)
    
    train_mse = mean_squared_error(y_train, model.predict(X_train_poly))
    test_mse = mean_squared_error(y_test, model.predict(X_test_poly))
    
    diag = "High Bias" if deg == 1 else ("Optimal" if deg == 3 else "High Variance (Overfit!)")
    print(f"  {deg:<8} | {train_mse:<12.4f} | {test_mse:<12.4f} | {diag}")
    
    # Plot curve
    ax = axes[i]
    ax.scatter(x_train, y_train, color='blue', alpha=0.7, label='Train Data')
    ax.scatter(x_test, y_test, color='red', alpha=0.7, label='Test Data')
    ax.plot(x_plot, model.predict(X_plot_poly), color='green', lw=2, label='Model Fit')
    ax.plot(x_plot, np.sin(x_plot), color='black', linestyle=':', label='True Sine')
    ax.set_ylim(-2.0, 2.0)
    ax.set_title(titles[i], fontsize=11, fontweight='bold')
    ax.grid(True, alpha=0.3)
    if i == 0:
        ax.legend(loc='lower left', fontsize=8)

plt.tight_layout()
fig_path = os.path.join(output_dir, "assets", "bias_variance_degrees.png")
fig.savefig(fig_path, dpi=120)
plt.close(fig)

print(f"\nVisualization successfully saved to: {fig_path}")
````

## 3. Train / Validation / Test Splits — Why and How

---

### 🗣️ In Simple English

When building a model, dividing your dataset into two pieces (**Train and Test**) is **not enough**. You need **three separate partitions**:

1. **Training Set (~70%):** The study material. The model updates its internal weights based on these examples.
2. **Validation Set (~15%):** The practice quizzes. You use this to compare different algorithms, tune hyperparameters (like learning rates, tree depths, and regularization), and decide when to stop training.
3. **Test Set (~15%):** The final locked exam. It stays in a sealed envelope until the very end. You evaluate your final champion model on it **exactly once** to report an honest, unbiased score.

Think of studying for university finals:
- **Train:** Homework assignments (you check answers and learn from mistakes).
- **Validation:** Mock midterm exams (you realize your study strategy needs adjustment).
- **Test:** The final state board exam (determines whether you get your license).

The key point is: **If you tune your model's hyperparameters by looking at the test set, you "leak" test information into your design decisions. The model overfits to the test set, creating a dangerous illusion of real-world accuracy.**

---

### 🌍 Real-Life Example

**Scenario:** Building an automated loan approval model for a commercial bank.

| Dataset Split | Partition | Who Uses It | What It Decides | Dangerous Mistake |
|---------------|:---------:|-------------|-----------------|-------------------|
| **Training Set** | 70% | Model Optimizer | Updates weights ($W$) | Memorizing customer names |
| **Validation Set** | 15% | Data Scientist / Engineer | Chooses model type (XGBoost vs Neural Net) & learning rate | Over-tuning until validation score is high |
| **Test Set** | 15% | Risk & Compliance Officer | Evaluates real-world default probability before launch | Peeking at test scores during development |

---

### 📖 How It Works (Step by Step)

1. **Stratified Splitting (Classification):**
   - If a disease occurs in only 2% of patients, a simple random split could put 0 sick patients into your test set!
   - **Stratified Split** preserves the exact class ratio across Train, Validation, and Test splits.
2. **Two-Step Splitting Protocol:**
   - Step 1: Split full dataset into `(Train+Val, Test)` with an 85/15 ratio.
   - Step 2: Split the remaining `Train+Val` into `(Train, Val)` so that final shares equal `70% / 15% / 15%`.
3. **Data Leakage Quarantine Rule:**
   - All feature scalers (`StandardScaler`), missing value imputers, and encoders must be **fitted on the Training set only**!
   - Then apply `.transform()` on Validation and Test. Never call `.fit()` or `.fit_transform()` on Validation or Test data!

---

### 🔑 Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Data Leakage** | Contaminating training pipelines with validation/test information | Yields falsely stellar metrics that collapse in production |
| **Stratified Sampling** | Preserving class ratios across all partitions | Vital for imbalanced classification tasks |
| **Golden Vault Rule** | Evaluating test data only once at the conclusion of a project | Guarantees true unbiased estimation of real-world generalization |

---

### 🗺️ Visual Overview

```
The Three-Way Partition Flow:

  Full Dataset (100%)
  =============================================================================
  [                           85%                           ] [   15% Test    ]  (Step 1: Vault Test Set!)
  =========================================================== =================
  [        70% Train         ] [       15% Val        ]
  ============================ =======================
       Updates Weights            Tunes Hyperparameters        Evaluated ONCE
      (Backpropagation)          (Early Stopping & Depth)      (Reported Metric)
```

---

### 🚫 Common Misconceptions

| ❌ Myth | ✅ Reality |
|--------|-----------|
| If you don't have enough data, a Train/Test split is fine. | If you tune hyperparameters on the test set, your test score is no longer an honest estimate of generalization. If data is limited, use **K-Fold Cross-Validation**! |

---

### 📊 Section Quick Reference

| Split | Share | Role | How Many Times Evaluated? |
|-------|:-----:|------|:-------------------------:|
| **Train** | 70% | Fit weights & parameters | Every batch / epoch |
| **Validation** | 15% | Hyperparameter tuning & model selection | Every epoch |
| **Test** | 15% | Final generalization estimate | Exactly ONCE |

````python
# ✅ Code Demo: Clean Leak-Free 70/15/15 Stratified Split on Iris
# Concept: Preserving Class Ratios and Preventing Preprocessing Leakage
# --------------------------------------------------------------------

# Load Iris dataset (150 samples across 3 classes: 50 each)
iris_path = "data/iris.csv" if os.path.exists("data/iris.csv") else "Module-04-Machine-Learning/data/iris.csv"
iris_df = pd.read_csv(iris_path)

X = iris_df.drop(columns=['target']).to_numpy()
y = iris_df['target'].to_numpy()

# Step 1: Split into Train+Val (85%) and Test (15%) with Stratification
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, test_size=0.15, stratify=y, random_state=42
)

# Step 2: Split Train+Val into Train (70% of total) and Val (15% of total)
# 15 / 85 = ~0.1765 to get equal 15% val split
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=(15/85), stratify=y_train_val, random_state=42
)

print("Stratified 70 / 15 / 15 Partition Counts:")
print(f"  Training Set:   {X_train.shape[0]} samples ({X_train.shape[0]/len(X)*100:.1f}%)")
print(f"  Validation Set: {X_val.shape[0]} samples ({X_val.shape[0]/len(X)*100:.1f}%)")
print(f"  Test Set:       {X_test.shape[0]} samples ({X_test.shape[0]/len(X)*100:.1f}%)")

print("\nClass Ratio Verification (Checking Stratification Integrity):")
print(f"  Train Class Counts: {np.bincount(y_train)} (Exact equal 33.3% share)")
print(f"  Val Class Counts:   {np.bincount(y_val)} (Exact equal 33.3% share)")
print(f"  Test Class Counts:  {np.bincount(y_test)} (Exact equal 33.3% share)")
print("  --> Stratification guaranteed that no class was underrepresented in any split!")
````

## 4. Cross-Validation — K-Fold & Stratified K-Fold

---

### 🗣️ In Simple English

What if you have a dataset with only 300 samples?
If you split it into 70/15/15, your validation set has only 45 samples. A single lucky or unlucky split could make your validation score swing wildly by $\pm 10\%$.

**$K$-Fold Cross-Validation** solves this by rotating the validation set across the entire dataset:
1. Chop your data into $K$ equal buckets (typically $K=5$).
2. Run 5 rounds of training. In each round, 4 buckets train the model and 1 bucket acts as the exam.
3. Every single sample gets tested exactly once!
4. Average all 5 test scores together to get a rock-solid, reliable evaluation.

The key point is: **Cross-validation eliminates split luck and provides an error bar ($\text{Accuracy} = 88.5\% \pm 1.8\%$) so you know how reliably your model will perform in the real world.**

---

### 🌍 Real-Life Example

**Scenario:** Auditing an algorithm across clinical drug trials.

| Round | Training Data | Validation Bucket | Validation Accuracy |
|:-----:|:-------------:|:-----------------:|:-------------------:|
| **Fold 1** | Buckets 2, 3, 4, 5 | Bucket 1 | $91.2\%$ |
| **Fold 2** | Buckets 1, 3, 4, 5 | Bucket 2 | $88.5\%$ |
| **Fold 3** | Buckets 1, 2, 4, 5 | Bucket 3 | $89.0\%$ |
| **Fold 4** | Buckets 1, 2, 3, 5 | Bucket 4 | $86.4\%$ |
| **Fold 5** | Buckets 1, 2, 3, 4 | Bucket 5 | $90.1\%$ |
| **Mean** | — | — | **$89.0\% \pm 1.6\%$** |

---

### 📖 How It Works (Step by Step)

1. **Partitioning:** The training dataset is randomly partitioned into $K$ equal folds.
2. **Stratification (Stratified $K$-Fold):** Ensures that every fold contains the exact same percentage of each target class as the complete dataset.
3. **Training Iterations:**
   For $k = 1 \dots K$:
   - Hold out fold $k$ as the validation fold.
   - Train on the remaining $K-1$ folds.
   - Evaluate on fold $k$ and record score $s_k$.
4. **Summary Aggregation:**
   $$\mu_{\text{score}} = \frac{1}{K} \sum_{k=1}^K s_k, \quad \sigma_{\text{score}} = \sqrt{\frac{1}{K} \sum_{k=1}^K (s_k - \mu)^2}$$
5. **Out-of-Fold (OOF) Predictions:**
   Because every sample is validated once when its fold is held out, you get clean out-of-fold predictions for your entire dataset, which are used to train meta-models (Stacking).

---

### 🔑 Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **$K$-Fold CV** | Rotating validation set across $K$ chunks | Maximizes data efficiency on small-to-medium datasets |
| **Stratified $K$-Fold** | Preserving class percentages within every fold | Prevents folds missing minority classes |
| **Confidence Interval** | Mean $\pm$ Standard Deviation of scores | Quantifies real-world uncertainty in model predictions |
| **OOF Predictions** | Out-Of-Fold predictions across all data points | Standard foundation for Kaggle-winning stacked ensemble models |

---

### 🗺️ Visual Overview

```
5-Fold Cross-Validation Workflow:

Iteration 1:  [  VAL  ] [ TRAIN ] [ TRAIN ] [ TRAIN ] [ TRAIN ]  -> Score 1
Iteration 2:  [ TRAIN ] [  VAL  ] [ TRAIN ] [ TRAIN ] [ TRAIN ]  -> Score 2
Iteration 3:  [ TRAIN ] [ TRAIN ] [  VAL  ] [ TRAIN ] [ TRAIN ]  -> Score 3
Iteration 4:  [ TRAIN ] [ TRAIN ] [ TRAIN ] [  VAL  ] [ TRAIN ]  -> Score 4
Iteration 5:  [ TRAIN ] [ TRAIN ] [ TRAIN ] [ TRAIN ] [  VAL  ]  -> Score 5
                                                                     ||
                                                Final Estimate: Mean +/- Std
```

---

### 🚫 Common Misconceptions

| ❌ Myth | ✅ Reality |
|--------|-----------|
| You must deploy all 5 models from 5-fold CV to production. | You can ensemble the 5 models, but standard practice is to use CV to find optimal hyperparameters, and then retrain a single final model on 100% of the training data! |

---

### 📊 Section Quick Reference

| Parameter | Recommended Value | Reason |
|-----------|:-----------------:|--------|
| **Folds ($K$)** | $K=5$ or $K=10$ | Optimal balance between bias, variance, and compute cost |
| **Stratified?** | Always for classification | Protects class distribution in every fold |
| **Shuffle?** | `shuffle=True` | Breaks arbitrary ordering in original CSV rows |

````python
# ✅ Code Demo: 5-Fold Stratified Cross-Validation on Iris Dataset
# Concept: Eliminating Split Luck and Computing Mean Performance with Uncertainty
# --------------------------------------------------------------------------------

# 5-Fold Stratified Splitter
k_folds = 5
skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=42)

fold_accuracies = []
oof_predictions = np.zeros(len(X))

print(f"Running {k_folds}-Fold Stratified Cross-Validation:")
print("-" * 50)

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
    X_train_fold, X_val_fold = X[train_idx], X[val_idx]
    y_train_fold, y_val_fold = y[train_idx], y[val_idx]
    
    # Train model on 4 folds
    clf = LogisticRegression(max_iter=500)
    clf.fit(X_train_fold, y_train_fold)
    
    # Evaluate on held-out fold
    preds = clf.predict(X_val_fold)
    oof_predictions[val_idx] = preds
    
    acc = accuracy_score(y_val_fold, preds)
    fold_accuracies.append(acc)
    print(f"  Fold {fold} | Validation Accuracy: {acc*100:5.2f}% ({len(val_idx)} samples)")

mean_acc = np.mean(fold_accuracies) * 100
std_acc = np.std(fold_accuracies) * 100

print("-" * 50)
print(f"--> Final 5-Fold Cross-Validation Result: {mean_acc:.2f}% (+/- {std_acc:.2f}%)")
print(f"--> Total Out-Of-Fold Accuracy across all 150 samples: {accuracy_score(y, oof_predictions)*100:.2f}%")
````

## 5. Evaluation Metrics: Accuracy, Precision, Recall, F1 & ROC-AUC

---

### 🗣️ In Simple English

Accuracy is often the most dangerous and misleading metric in machine learning:
If a bank processes 1,000,000 transactions and only 100 are fraudulent (0.01%), a dumb program that labels **every single transaction as safe** achieves **99.99% accuracy**. Yet it catches **zero fraudsters**!

To truly measure performance, we use specialized metrics:
- **Accuracy:** Overall percentage of correct predictions. (Only useful when classes are balanced!).
- **Precision:** *"When the model predicts Positive, how often is it right?"* (Focuses on minimizing false alarms).
- **Recall (Sensitivity):** *"Out of all real Positive cases out there, how many did the model catch?"* (Focuses on not missing true cases).
- **F1-Score:** The harmonic balance between Precision and Recall.
- **ROC-AUC:** Measures how well the model ranks true positives higher than negatives across every possible confidence threshold.

---

### 🌍 Real-Life Example

**Scenario:** Spam filter vs Cancer screening.

| Application | Cost of False Positive | Cost of False Negative | Prioritize | Why |
|-------------|------------------------|------------------------|:----------:|-----|
| **Spam Filter** | High (Critical business email sent to spam folder) | Low (A stray spam email reaches inbox) | **Precision** | False alarms ruin business operations |
| **Cancer Screening** | Low (Patient attends a harmless follow-up MRI) | Fatal (Malignant tumor goes undetected) | **Recall** | Must catch 100% of true cases at all costs |
| **Customer Churn** | Moderate (Discount coupon sent to happy customer) | Moderate (Valuable customer leaves platform) | **F1-Score** | Balances retention cost against churn risk |

---

### 📖 How It Works (Step by Step)

1. **Confusion Matrix Building Blocks:**
   - **True Positives (TP):** Reality = 1, Prediction = 1
   - **True Negatives (TN):** Reality = 0, Prediction = 0
   - **False Positives (FP):** Reality = 0, Prediction = 1 (False alarm)
   - **False Negatives (FN):** Reality = 1, Prediction = 0 (Missed case)
2. **Mathematical Formulations:**
   $$\text{Precision} = \frac{\text{TP}}{\text{TP} + \text{FP}}, \quad \text{Recall} = \frac{\text{TP}}{\text{TP} + \text{FN}}$$
   $$\text{F1-Score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$
3. **ROC Curve (Receiver Operating Characteristic):**
   - A model outputs a continuous probability score $p \in [0, 1]$.
   - Instead of picking a single threshold like $0.5$, sweep threshold $\tau$ from $1.0$ down to $0.0$.
   - Plot **True Positive Rate (Recall)** on the $y$-axis against **False Positive Rate ($\frac{\text{FP}}{\text{FP} + \text{TN}}$)** on the $x$-axis.
4. **ROC-AUC (Area Under the Curve):**
   - Measures the total two-dimensional area under the ROC curve.
   - $\text{AUC} = 0.5$: Worthless (random coin toss).
   - $\text{AUC} = 1.0$: Flawless ranking (all positives ranked above all negatives).

---

### 🔑 Key Concepts

| Concept | What It Means | Why It Matters |
|---------|---------------|----------------|
| **Accuracy Paradox** | High accuracy achieved by always guessing the majority class | Proves accuracy is useless under severe class imbalance |
| **Precision-Recall Tradeoff** | Increasing precision lowers recall, and vice-versa | Engineering choice determined by business cost of false alarms vs misses |
| **F1-Score** | Harmonic mean of Precision & Recall | Fails if either metric collapses; ideal for single-number evaluation |
| **ROC-AUC** | Probability that a random positive sample ranks higher than a random negative | Threshold-independent ranking quality metric |

---

### 🗺️ Visual Overview

```
ROC Curve & Precision-Recall Tension:

  ROC Curve:                               Precision-Recall Tradeoff:
  True Positive Rate (Recall)               Precision ^
  1.0 ^         ______* Ideal Curve (AUC=0.95)       |   \
      |       _/'                                    |    \  Tradeoff curve
      |     _/'                                      |     \
      |   _/'   Random Guess (AUC=0.50)              |      \
      |  / - - - - - - -                             +------------------> Recall
  0.0 +--------------------------> FPR              0.0                 1.0
     0.0                      1.0
```

---

### 🚫 Common Misconceptions

| ❌ Myth | ✅ Reality |
|--------|-----------|
| The classification decision threshold must always be $0.5$. | In medical diagnosis, you lower the threshold to $0.1$ to maximize Recall. In spam filtering, you raise it to $0.9$ to maximize Precision! |

---

### 📊 Section Quick Reference

| Metric | When to Prioritize | Formula |
|--------|--------------------|---------|
| **Precision** | Cost of False Alarm is high | $\frac{\text{TP}}{\text{TP} + \text{FP}}$ |
| **Recall** | Cost of Missing a Target is high | $\frac{\text{TP}}{\text{TP} + \text{FN}}$ |
| **F1-Score** | Imbalanced data needing a single balance | $2 \frac{P \cdot R}{P + R}$ |
| **ROC-AUC** | Comparing threshold-independent ranking quality | Area under TPR vs FPR curve |

````python
# ✅ Code Demo: Evaluating Imbalanced Predictions & ROC-AUC Curves
# Concept: Demonstrating the Accuracy Paradox, F1-Score & ROC Threshold Sweeps
# ----------------------------------------------------------------------------

# Simulate imbalanced medical scenario: 1000 patients, only 50 are sick (5% positive rate)
np.random.seed(42)
n_samples = 1000
y_true = np.zeros(n_samples, dtype=int)
y_true[:50] = 1  # 50 positives

# Simulated model probabilities (noisy prediction with decent signal)
y_probs = np.random.beta(0.5, 5.0, size=n_samples)
y_probs[y_true == 1] += np.random.uniform(0.3, 0.6, size=50)
y_probs = np.clip(y_probs, 0.0, 1.0)

# Binary predictions using standard default threshold = 0.5
y_pred_default = (y_probs >= 0.5).astype(int)

# 1. Compute Metrics
acc  = accuracy_score(y_true, y_pred_default)
prec = precision_score(y_true, y_pred_default)
rec  = recall_score(y_true, y_pred_default)
f1   = f1_score(y_true, y_pred_default)
auc  = roc_auc_score(y_true, y_probs)

print("Imbalanced Medical Screening Evaluation (Default Threshold = 0.5):")
print(f"  Accuracy:  {acc*100:5.2f}% (Sounds great, but misleading due to 95% negative base!)")
print(f"  Precision: {prec*100:5.2f}% (Out of flagged patients, {prec*100:.1f}% were truly sick)")
print(f"  Recall:    {rec*100:5.2f}% (Model caught {rec*100:.1f}% of all sick patients)")
print(f"  F1-Score:  {f1*100:5.2f}% (Harmonic balance)")
print(f"  ROC-AUC:   {auc:6.4f} (Threshold-independent ranking quality)")

# 2. Sweep ROC Curve
fpr, tpr, thresholds = roc_curve(y_true, y_probs)

fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(fpr, tpr, color='#1f77b4', lw=2.5, label=f'Model ROC (AUC = {auc:.3f})')
ax.plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random Chance (AUC = 0.500)')
ax.set_title("Receiver Operating Characteristic (ROC) Curve", fontsize=13, fontweight='bold')
ax.set_xlabel("False Positive Rate (1 - Specificity)", fontsize=11)
ax.set_ylabel("True Positive Rate (Recall / Sensitivity)", fontsize=11)
ax.grid(True, alpha=0.3)
ax.legend(loc='lower right', fontsize=11)

plt.tight_layout()
fig_path = os.path.join(output_dir, "assets", "roc_curve_medical.png")
fig.savefig(fig_path, dpi=120)
plt.close(fig)

print(f"\nROC Curve plot successfully saved to: {fig_path}")
````
